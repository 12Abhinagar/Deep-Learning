```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Bidirectional, SimpleRNN, Dense,LSTM,GRU
```

```python
# Load the IMDb dataset
num_words = 10000
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=num_words)

# Pad sequences to have the same length
max_len =100
x_train = pad_sequences(x_train, maxlen=max_len,padding='post', truncating='post')
x_test = pad_sequences(x_test, maxlen=100,padding='post', truncating='post')

embedding_dim=32
```

```python
model = Sequential([
    Embedding(input_dim=num_words,output_dim=embedding_dim,input_length=max_len),
    SimpleRNN(5),
    Dense(1,activation='sigmoid')
])

model.build(input_shape=(None,max_len))

model.summary()
```

```python
# Bidirectional RNN model
model1 = Sequential([
    Embedding(input_dim=num_words,output_dim=embedding_dim,input_length=max_len),
    Bidirectional(SimpleRNN(5)),
    Dense(1,activation='sigmoid')
])

model1.build(input_shape=(None,max_len))

model1.summary()
```

```python
# Bidirectional LSTM model
model2 = Sequential([
    Embedding(input_dim=num_words,output_dim=embedding_dim,input_length=max_len),
    Bidirectional(LSTM(5)),
    Dense(1,activation='sigmoid')
])

model2.build(input_shape=(None,max_len))

model2.summary()
```

```python
# Bidirectional GRU model
model3 = Sequential([
    Embedding(input_dim=num_words,output_dim=embedding_dim,input_length=max_len),
    Bidirectional(GRU(5)),
    Dense(1,activation='sigmoid')
])

model3.build(input_shape=(None,max_len))

model3.summary()
```
